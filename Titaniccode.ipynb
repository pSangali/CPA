```python
import pandas as pd

# Carregar o CSV
df = pd.read_csv("titanic.csv")

# Verificar ausentes
print("Valores ausentes por coluna:")
print(df.isnull().sum())

# Preenche a coluna 'Age' (idade) com a mediana
df['Age'].fillna(df['Age'].median(), inplace=True)

# Preenche a coluna 'Cabin' com 'Desconhecido'
df['Cabin'].fillna('Desconhecido', inplace=True)

# Preencher a coluna 'Embarked' com o valor mais frequente (moda)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# Verificar se ainda há valores ausentes
print("\nValores ausentes após o preenchimento:")
print(df.isnull().sum())

# Salvar o novo CSV
df.to_csv("titanic_tratado.csv", index=False)

print("\nArquivo 'titanic_tratado.csv' salvo com sucesso!")
```
